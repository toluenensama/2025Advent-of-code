```python
%%bash
set -euo pipefail

MAMBA_ROOT="$HOME/mamba"
MAMBA_BIN="$MAMBA_ROOT/bin/micromamba"
if [ ! -x "$MAMBA_BIN" ]; then
  mkdir -p "$MAMBA_ROOT"
  curl -Ls https://micro.mamba.pm/api/micromamba/linux-64/latest | tar -xvj -C "$MAMBA_ROOT" bin/micromamba >/dev/null
fi

"$MAMBA_BIN" create -y -r "$MAMBA_ROOT" -n psi4 -c conda-forge psi4 python=3.12 >/dev/null
```

```python
import re

# Read the content of psi4.out
with open('psi4.out', 'r') as f:
    output_content = f.read()

# Regular expression to find 'CCSD(T) total energy' and the corresponding float value
# The pattern looks for the exact phrase and then captures the floating-point number
pattern = re.compile(r"CCSD\(T\) total energy\s+=\s+(-?\d+\.\d+)")

# Find all occurrences of the pattern
ccsd_t_energies = pattern.findall(output_content)

# Print the extracted energies
if ccsd_t_energies:
    print("Extracted CCSD(T) Total Energies:")
    for energy in ccsd_t_energies:
        print(energy)
else:
    print("No CCSD(T) total energies found in the output.")
```
